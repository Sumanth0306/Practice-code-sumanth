Business Rules
Minimum cart value must be reached.
Calculate the percentage discount.
Compare the calculated discount with the maximum discount allowed.
Apply whichever discount rule is appropriate.
If the minimum cart value is not reached, tell the customer how much more needs to be added.
Algorithm / Pseudocode
Create a program to apply discount of 10% if the min amt is 600 and max cap of discount is 100 rupees.

Identify the variables - min_cart_amt, disc_pct_whole, max_disc_amt
Identify the input (runtime) - cart_value
First check: if cart_value >= min_cart_amt then i am going to apply offer else i am going notify user to add some more amout to get offer. min_cart_amt - cart_value
Calculate discount in percentage (fraction): disc_pct = 10/100
Calculate discount amount after applying to the cart amt, if the discount amt > max discount, then reduce max discount to the cart amount and produce final amount, otherwise reduce calculated discount amt to the cart amount and produce final amount.

```python
#Nested (multiple levels of looping) Looping Constructs
days1=['mon','tue','wed','thu','fri','sat','sun']
batches1=['wd38','we49']
for batch in batches1:
    print(f" execute batch {batch} ")
    if batch == 'wd38':
         for day in days1:
            if day in ('sat','sun'):
                break
            print(f"executing batch {batch} in day {day}")
    elif batch == 'we49':
         for day in days1:
            if day not in ('sat','sun'):
                continue
            print(f"executing batch {batch} in day {day}")
```
